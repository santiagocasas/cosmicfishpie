### Compare the Power Spectrum from two different cosmologies

```python
import numpy as np

def get_param_string_from_value(cosmopars, fiducialpars, param_names, folder_param_dict, eps_values, E_00=True):
    """
    Get the parameter string from the cosmological parameters.

    Parameters
    -----------
    cosmopars : dict
        Dictionary of cosmological parameters.
    fiducialpars : dict
        Dictionary of fiducial cosmological parameters.
    param_names : list
        List of parameter names to check.
    folder_param_dict : dict
        Dictionary mapping parameter names to folder names.
    eps_values : list
        List of epsilon values to use for rounding.
    E_00 : bool, optional
        Whether to keep E-00 notation or replace with E- (default is True).

    Returns
    --------
    str
        Parameter string representing the input cosmological parameters.
    """
    rel_tol = 1e-5
    signstrings = {-1.0: "mn", 1.0: "pl"}
    
    for parname in param_names:
        if not np.isclose(cosmopars[parname], fiducialpars[parname], rtol=rel_tol):
            if fiducialpars[parname] != 0:
                delta_eps = (cosmopars[parname] / fiducialpars[parname]) - 1.0
            elif fiducialpars[parname] == 0.0:
                delta_eps = cosmopars[parname] - fiducialpars[parname]
            eps_sign = np.sign(delta_eps)
            sign_string = signstrings[eps_sign]
            
            allowed_eps_vals = np.concatenate((eps_values, -np.array(eps_values), np.array([0])))
            delta_eps = round(delta_eps, 6)  # Simplified rounding
            delta_eps = min(allowed_eps_vals, key=lambda x: abs(x-delta_eps))
            
            eps_string = "{:.1E}".format(abs(delta_eps))
            eps_string = eps_string.replace(".", "p")
            if not E_00:
                eps_string = eps_string.replace("E-0", "E-")
            folder_parname = folder_param_dict[parname]
            param_folder_string = (
                folder_parname + "_" + sign_string + "_" + "eps" + "_" + eps_string
            )
            return param_folder_string
    
    return "fiducial_eps_0"

# Example usage:
cosmopars = {'Omegam': 0.32*(1+0.0125*3), 'Omegab': 0.05, 'h': 0.67, 'ns': 0.96, 'sigma8': 0.815584}
fiducialpars = {'Omegam': 0.32, 'Omegab': 0.05, 'h': 0.67, 'ns': 0.96, 'sigma8': 0.815584}
param_names = ['Omegam', 'Omegab', 'h', 'ns', 'sigma8']
folder_param_dict = {'Omegam': 'Om', 'Omegab': 'Ob', 'h': 'h', 'ns': 'ns', 'sigma8': 's8'}
eps_values = [0.00625, 0.01, 0.0125, 0.01875, 0.025, 0.03, 0.0375, 0.05, 0.1]

result = get_param_string_from_value(cosmopars, fiducialpars, param_names, folder_param_dict, eps_values)
print(result)
```
