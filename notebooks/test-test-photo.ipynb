```python
from cosmicfishpie.likelihood import PhotometricLikelihood
from cosmicfishpie.fishermatrix import cosmicfish
import numpy as np
```

```python
def photometric_fisher_matrix():
    options = {
        "accuracy": 1,
        "feedback": 1,
        "code": "symbolic",
        "outroot": "test_photo_like",
        "survey_name": "Euclid",
        "survey_name_photo": "Euclid-Photometric-ISTF-Pessimistic",
        "survey_name_spectro": False,
        "specs_dir": "../cosmicfishpie/configs/default_survey_specifications/",
        "cosmo_model": "LCDM",
    }

    fiducial = {
        "Omegam": 0.3145714273,
        "Omegab": 0.0491989,
        "h": 0.6737,
        "ns": 0.96605,
        "sigma8": 0.81,
        "w0": -1.0,
        "wa": 0.0,
        "mnu": 0.06,
        "Neff": 3.044,
    }

    observables = ["GCph"]

    return cosmicfish.FisherMatrix(
        fiducialpars=fiducial,
        options=options,
        observables=observables,
        cosmoModel=options["cosmo_model"],
        surveyName=options["survey_name"],
    )
```

```python
cosmoFM_fid  = photometric_fisher_matrix()
```

```python
def _sample_params():
    samp_pars = {
    'Omegam': 0.3145714273,
    'Omegab': 0.0491989,
    'h': 0.6737,
    'ns': 0.96605,
    'sigma8': 0.81,
    'w0': -1.0,
    'wa': 0.0,
    'mnu': 0.06,
    'Neff': 3.044,
    'bias_model': 'binned',
    'b1': 1.0997727037892875,
    'b2': 1.220245876862528,
    'b3': 1.2723993083933989,
    'b4': 1.316624471897739,
    'b5': 1.35812370570578,
    'b6': 1.3998214171814918,
    'b7': 1.4446452851824907,
    'b8': 1.4964959071110084,
    'b9': 1.5652475842498528,
    'b10': 1.7429859437184225,
    'fout': 0.1,
    'co': 1,
    'cb': 1,
    'sigma_o': 0.05,
    'sigma_b': 0.05,
    'zo': 0.1,
    'zb': 0.0,
    'IA_model': 'eNLA',
    'AIA': 1.72,
    'betaIA': 2.17,
    'etaIA': -0.41*1.1
    }
    return samp_pars

sampled_pars = _sample_params()
```

```python

photo_like = PhotometricLikelihood(
    cosmoFM_data=cosmoFM_fid,
    cosmoFM_theory=cosmoFM_fid,
    observables=cosmoFM_fid.observables,
)

target_ell = 300  # adjust as needed
ells = photo_like.data_obs["ells"]
idx = int(np.argmin(np.abs(ells - target_ell)))
ell_value = ells[idx]

data_cells = photo_like.data_obs
theory_cells = photo_like.compute_theory(dict(sampled_pars))

print(f"ℓ ≈ {ell_value}")
print("data Cell_GG row (n_GCph × n_GCph):")
print(data_cells["Cell_GG"][idx][1,8])
print("theory Cell_GG row (n_GCph × n_GCph):")
print(theory_cells["Cell_GG"][idx][1,8])
print("lmax_GCph:", photo_like.cosmoFM_data.specs["lmax_GCph"])
print("fsky_GCph:", photo_like.photo_cov_data.fsky_GCph)
print("ell sampling", photo_like.cosmoFM_data.specs.get("ell_sampling"))
print("loglike:", photo_like.loglike(param_dict=sampled_pars))
```
